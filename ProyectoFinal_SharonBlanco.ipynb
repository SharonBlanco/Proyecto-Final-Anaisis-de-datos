### Normalización de puestos

* `Analista BI` + `Analista de Datos BI` + `Analista de Datos de BI` → **Analista de Datos BI**

* `Ingeniero de ML` + `Ingeniero de Machine Learning` → **Ingeniero de Machine Learning**

* `Analista Líder de Datos` + `Analista de Datos Líder` →  **Analista de Datos Líder**

---

### Normalización de países

* `TÃ¼rkiye` → **Türkiye**

* `Korea, Republic of` → **South Korea**

* `Iran, Islamic Republic of` → **Iran**

* `Moldova, Republic of` → **Moldova**

* `Bolivia, Plurinational State of` → **Bolivia**

* `Viet Nam` → **Vietnam**

* `Russian Federation` → **Russia**



```python
# Corrección de nombres de países
correccion_paises = {
    'TÃ¼rkiye': 'Türkiye',
    'Korea, Republic of': 'South Korea',
    'Iran, Islamic Republic of': 'Iran',
    'Moldova, Republic of': 'Moldova',
    'Bolivia, Plurinational State of': 'Bolivia',
    'Viet Nam': 'Vietnam',
    'Russian Federation': 'Russia'
}

# Aplicar correcciones en ambas columnas, se reemplaza con el valor del diccionario
for columna in ['pais_empleado', 'pais_empresa']:
    df[columna] = df[columna].replace(correccion_paises)
```

```python
#Pequeña prueba para visualizar si se aplicaron los cambios correctamente
df[df.puesto == 'Ingeniero de ML']
```

```python
# Corrección de puestos duplicados por escritura diferente
correccion_puestos = {
    'Analista BI': 'Analista de Datos BI',
    'Analista de Datos de BI': 'Analista de Datos BI',
    'Ingeniero de ML': 'Ingeniero de Machine Learning',
    'Ingeniero de ML Ops': 'Ingeniero MLOps',
    'Ingeniero de Operaciones de ML': 'Ingeniero MLOps',
    'Desarrollador BI': 'Desarrollador de Inteligencia de Negocios',
    'Analista Líder de Datos': 'Analista de Datos Líder'
}

# Aplicar correcciones
df['puesto'] = df['puesto'].replace(correccion_puestos)
```

# Análisis EDA

### ¿Cómo varía el salario según el nivel de experiencia?



```python
#Se calcula el salario promedio por experiencia
df.groupby('nivel_experiencia').salario_usd.mean().sort_values(ascending= False)
```

Visualizando el promedio de salario en cada categoría de nivel de experiencia, Ejecutivo tiene el salario más alto y Junior el más bajo, estos resultados presentan que entre más alta la experiencia y más cercanía hay hacia administración, más aumenta el salario.

### ¿El trabajo remoto paga mejor, igual o peor que el presencial?


```python
# Se calcula el salario promedio por modalidad
df.groupby('modalidad').salario_usd.mean().sort_values(ascending= False)
```

El trabajo remoto paga menos que el presencial, pero más que el híbrido.

### ¿Qué categoría de puesto es la mejor pagada?


```python
# Se calcula el salario promedio por categoría y se toma solo el primero
df.groupby('categoria_puesto').salario_usd.mean().sort_values(ascending= False).head(1)
```

La categoría mejor pagada es la de Machine Learning e IA, la razón se debe a que los ingenieros de Machine Learning tiene que tener tanto conocimiento matemático como en arquitectura de software y programación, lo cual son dos habilidades que resultan dificiles de dominar.

### ¿Cómo ha cambiado el salario promedio entre 2020 y 2024?


```python
# Se calcula el salario promedio por año
df.groupby('anio').salario_usd.mean().sort_index(ascending=False)
```

El salario promedio ha ido aumentando desde el 2020 hasta el 2023 pero en 2024 ha disminuido un poco.

### ¿En qué país se concentran más las ofertas de empleo?


```python
# Se calcula el país con más empleos
df.pais_empresa.value_counts().head(1)
```

Las ofertas de empleo se concentran más en Estados Unidos, con un total de 7215 registros de trabajadores.

### ¿El tamaño de la empresa influye en el salario?


```python
#Se agrupa por el tamaño de la empresa y se calcula el promedio del salario
df.groupby('tamanio_empresa').salario_usd.mean().sort_values(ascending= False)
```

No, ya que si influyera, la empresa grande tendría mayor salario promedio y no es el caso. Según la base de datos la empresa que ofrece mayor salario es la empresa mediana.

### ¿Qué tipo de empleo (Tiempo completo, Freelance, Medio tiempo) predomina en el sector y cómo varía el salario entre ellos?

```python
#Se agrupa por tipo de empleo y se calculan varios cálculos estadísticos del salario
df.groupby('tipo_empleo').salario_usd.agg(['count', 'mean', 'median']).sort_values(by = 'mean', ascending = False)
```

El tipo de empleo que predomina en el sector corresponde al de tiempo completo, asimismo, es el que gana más en promedio. El tipo de empleo que gana menos corresponde al freelance, debido a que es un tipo de empleo en el cual se trabaja con clientes directamente que contratan el servicio, no es fijo, por ende, es el que gana en promedio menos.

### ¿Existe una brecha salarial o de oportunidades si la empresa está en un país diferente al del empleado? (Trabajo internacional vs. local)


```python
#Agrupar los empleados extrangeros y los empleados nacionales
empleado_extrangero = df[df.pais_empleado != df.pais_empresa]
empleado_nacional = df[df.pais_empleado == df.pais_empresa]


print('Empleado extrangero: ')
print(empleado_extrangero.salario_usd.agg(['count', 'mean', 'median']))
print()
print()
print('Empleado nacional: ')
print(empleado_nacional.salario_usd.agg(['count', 'mean', 'median']))
```

Sí hay una diferencia, el salario promedio de empleado nacional es más alto, esto puede deberse a varios factores, el primero es porque en el país de la empresa los empleados nacionales tienen sus derechos laborales, por lo cual las empresas a veces optan por contratar con modalidad remoto a extrangeros, a los cuales pueden disminuirles el salario. Asimismo, ciertos países tienen un precio de vida muy caro, y al contratar remoto, los extrangeros no toman mucho en cuenta eso, y aceptan el salario que las empresas les ofrecen, que por supuesto para el estilo de vida de su propio país es muy alto, pero no toman en cuenta que al trabajador nacional le pueden estar pagando el doble o triple.

###  ¿Cómo ha evolucionado la adopción de la modalidad de trabajo (Remoto, Presencial, Híbrido) año tras año (2020 vs 2024)?


```python
#Agrupar por año y modalidad
df.groupby(['anio', 'modalidad']).size()
```

Revisando las agrupaciones de datos, se puede tomar en cuenta que la cantidad de registro de datos ha ido aumentando, en 2020 la cantidad de registros es mucho menor que en 2024. Asimismo, se ha aumentado a lo largo de los años el trabajo presencial exponencialmente. Por otro lado, la adopción al trabajo remoto e híbrido ha ido aumentando linealmente solo que en 2024 disminuyó.

### Para los puestos Junior, ¿cuáles son las 3 categorías de puestos mejor pagadas y en qué países hay más oportunidades para ellos?

```python
# Se separan solo los puestos junior y se calcula por categoria de puestos y el salario
puestos_junior = df[df.nivel_experiencia == 'Junior']
puestos_junior.groupby(['categoria_puesto']).salario_usd.agg(['count', 'mean']).sort_values(by = 'mean', ascending = False).head(3)
```

```python
# Se agrupan los puestos por país, y se muestran los primeros 3 con mayor cantidad de empleados
puestos_junior.groupby(['pais_empresa']).size().sort_values(ascending = False).head(3)
```

Las categorías mejor pagadas para Junior son Ciencia de Datos e Investigacion, Machine Learning e IA e Ingenieria de Datos, y el país con mejor oportunidad laboral para junior es Estados Unidos.

### ¿Las empresas grandes prefieren una modalidad de trabajo distinta (ej. presencial o híbrida) en comparación con las empresas pequeñas y medianas?

```python
# Se agrupa la cantidad por tamaño de empresa y modalidad
df.groupby(['tamanio_empresa', 'modalidad']).size()
```

En la empresa grande, no se visualiza una diferencia muy marcada entre la cantidad de trabajadores de cada modalidad, sin embargo se presencian más puestos en modalidad remota. Algo similar pasa con la empresa pequeña. Sin embargo, en la empresa mediana, dominan los empleados en modalidad presencial, seguido de modalidad remota y por último la modalidad híbrida.

### ¿Cuáles son los 5 títulos de puestos específicos más frecuentes en el mercado laboral actual (2024)?


```python
#Se filtran los puestos por año y se cuentan los registros por puesto específico
puestos_2024 = df[df.anio == 2024]
puestos_2024.groupby('puesto').size().sort_values(ascending= False).head()
```

Los 5 puestos más frecuentes en el mercado laboral actual son Cientifico de Datos, Ingeniero de Datos, Analista de Datos, Ingeniero de Machine Learning y Ciencia de Datos.

### Para los puestos Senior, ¿el trabajo remoto sigue pagando bien o la brecha remoto/presencial se mantiene igual que en el análisis general?

# Visualizaciones

```python

# Se crea una fila con 3 columnas
fig, axes = plt.subplots(1, 3, figsize=(20, 6))
fig.suptitle('Análisis de Contratos, Modalidades y Tendencias Temporales', fontsize=16, fontweight='bold', y=1.05)

# --- GRÁFICO 1: Tipo de Empleo vs Salario ---
salario_tipo = df.groupby('tipo_empleo')['salario_usd'].mean().sort_values(ascending=False)
axes[0].bar(salario_tipo.index, salario_tipo.values, color=['#1F4E79', '#2E75B6', '#4BACC6'])
axes[0].set_title('Salario Promedio por Tipo de Empleo', fontweight='bold')
axes[0].set_ylabel('Salario USD')
axes[0].yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x/1000:.0f}k'))

# --- GRÁFICO 2: Evolución de Modalidad por Año  ---
# Usamos un gráfico de barras agrupadas con seaborn
sns.countplot(data=df, x='anio', hue='modalidad', palette='Blues_r', ax=axes[1])
axes[1].set_title('Evolución de Modalidades (2020-2024)', fontweight='bold')
axes[1].set_xlabel('Año')
axes[1].set_ylabel('Cantidad de Empleos')
axes[1].legend(title='Modalidad')

# --- GRÁFICO 3: Modalidad según el Tamaño de Empresa ---
# Un gráfico cruzado para ver las preferencias de las empresas
sns.countplot(data=df, x='tamanio_empresa', hue='modalidad', palette='viridis', ax=axes[2])
axes[2].set_title('Modalidad por Tamaño de Empresa', fontweight='bold')
axes[2].set_xlabel('Tamaño de Empresa')
axes[2].set_ylabel('Cantidad')
axes[2].legend(title='Modalidad')

plt.tight_layout()
plt.show()
```

En estos gráficos podemos visualizar de manera gráfica algunsa preguntas agregadas al estudio EDA, el salario promedio por tipo de empleo va en orden descendente desde tiempo completo hasta freelance. El segundo gráfico muestra el aumento de registros de datos desde el 2020 hasta el 2024, y como en 2023 y 2024 dominan los empleos presenciales. Por último, en el último gráfico se pueden visualizar las modalidades por empresa. Algo muy interesante es que la mayoría de trabajadores registrados trabajan en una empresa mediana, asimismo, domina el trabajo presencial sobre el remoto.

```python
# Se crea una fila con 3 columnas para los análisis avanzados
fig, axes = plt.subplots(1, 3, figsize=(22, 6))
fig.suptitle('Análisis de Roles, Brechas Internacionales y Mercado Junior', fontsize=16, fontweight='bold', y=1.05)

# --- GRÁFICO 4: Trabajo Extranjero vs Nacional  ---
# Se crea una columna rápida para identificar si es internacional
df['es_extranjero'] = df.apply(lambda r: 'Extranjero' if r['pais_empleado'] != r['pais_empresa'] else 'Nacional', axis=1)
salario_brecha = df.groupby('es_extranjero')['salario_usd'].mean()
axes[0].bar(salario_brecha.index, salario_brecha.values, color=['#4F81BD', '#C0504D'])
axes[0].set_title('Brecha Salarial: Nacional vs Extranjero', fontweight='bold')
axes[0].set_ylabel('Salario Promedio (USD)')
axes[0].yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x/1000:.0f}k'))

# --- GRÁFICO 5: Top 3 Categorías para Juniors ---
df_junior = df[df['nivel_experiencia'] == 'Junior']
top3_junior = df_junior.groupby('categoria_puesto')['salario_usd'].mean().sort_values(ascending=True).tail(3)
axes[1].barh(top3_junior.index, top3_junior.values, color='#2E75B6')
axes[1].set_title('Top 3 Categorías Mejor Pagadas (Junior)', fontweight='bold')
axes[1].set_xlabel('Salario Promedio (USD)')
axes[1].xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x/1000:.0f}k'))

# --- GRÁFICO 6: Top 5 Puestos más demandados en 2024 ---
df_2024 = df[df['anio'] == 2024]
top5_puestos = df_2024['puesto'].value_counts().head(5)
axes[2].bar(top5_puestos.index, top5_puestos.values, color='#1F4E79')
axes[2].set_title('Top 5 Puestos Más Demandados (2024)', fontweight='bold')
axes[2].set_xticks(range(len(top5_puestos.index)))
axes[2].set_xticklabels(top5_puestos.index, rotation=45, ha='right')
axes[2].set_ylabel('Frecuencia / Ofertas')

plt.tight_layout()
plt.show()
```

En este segundo panel, se puede visualizar en el primer gráfico que el trabajador nacional es mejor pagado que el extranjero si visualizamos el salario promedio. En el segundo gráfico se visualizan las 3 categorías mejor pagadas, las cuales corresponden a Ciencia de Datos e Investigación, Machine Learning e IA e Ingeniería de Datos. El el último gráfico se encuentran los 5 puestos más demandados en 2024: Cientifico de Datos, Ingeniero de Datos, Analista de Datos, Ingeniero de Machine Learning y Ciencia de Datos.
